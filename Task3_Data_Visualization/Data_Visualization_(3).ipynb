```python
import pandas as pd

df = pd.read_csv("books_scraped.csv")

print("Dataset loaded successfully!")
display(df.head())
```

STEP 3: Understand the Dataset

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nFirst 5 Records:")
display(df.head())
```

STEP 4: Visualize Book Prices

```python
plt.figure(figsize=(10, 8))

sorted_df = df.sort_values("Price", ascending=True)

plt.barh(sorted_df["Book Name"], sorted_df["Price"])

plt.xlabel("Price (£)")
plt.ylabel("Book Name")
plt.title("Book Prices")
plt.tight_layout()

plt.show()
```

STEP 5: Visualize Rating Distribution

```python
plt.figure(figsize=(8, 5))

rating_order = ["One", "Two", "Three", "Four", "Five"]

sns.countplot(
    x="Rating",
    data=df,
    order=rating_order
)

plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Ratings")

plt.show()
```

STEP 6: Compare Price and Rating

```python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating_Numeric"] = df["Rating"].map(rating_map)

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rating_Numeric",
    y="Price",
    s=100
)

plt.xlabel("Rating")
plt.ylabel("Price (£)")
plt.title("Relationship Between Book Rating and Price")
plt.xticks([1, 2, 3, 4, 5])

plt.show()
```

STEP 7: Analyze Price Distribution

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Price"], bins=8, kde=True)

plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Prices")

plt.show()
```

STEP 8: Create Correlation Heatmap

```python
correlation = df[["Price", "Rating_Numeric"]].corr()

plt.figure(figsize=(6, 4))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Price and Rating Correlation")
plt.show()
```

STEP 9: Summarize Visual Insights

```python
print("VISUAL INSIGHTS")
print("-" * 40)

highest_price_book = df.loc[df["Price"].idxmax(), "Book Name"]
lowest_price_book = df.loc[df["Price"].idxmin(), "Book Name"]
most_common_rating = df["Rating"].mode()[0]

print("1. Highest-priced book:", highest_price_book)
print("2. Highest price:", df["Price"].max())
print("3. Lowest-priced book:", lowest_price_book)
print("4. Lowest price:", df["Price"].min())
print("5. Average book price:", round(df["Price"].mean(), 2))
print("6. Most common rating:", most_common_rating)
print("7. Total number of books:", len(df))
```
