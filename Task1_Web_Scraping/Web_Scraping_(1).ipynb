STEP 1: libraries import


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

print("Libraries imported successfully!")
```

STEP 2: Website URL + Request

```python
url = "https://books.toscrape.com/"

response = requests.get(url)

print("Status Code:", response.status_code)
print("Website connected successfully!")
```

STEP 3 — HTML content extract

```python
soup = BeautifulSoup(response.text, "html.parser")

print("HTML content extracted successfully!")
print(soup.title.text)
```

STEP 4 — Book names extract

```python
books = soup.find_all("article", class_="product_pod")

book_names = []

for book in books:
    name = book.h3.a["title"]
    book_names.append(name)

print("Number of books found:", len(book_names))
print("\nFirst 5 books:")

for name in book_names[:5]:
    print(name)
```

STEP 5 — Price + Rating extract

```python
book_data = []

for book in books:
    name = book.h3.a["title"]
    price = book.find("p", class_="price_color").text.strip()
    rating = book.find("p", class_="star-rating")["class"][1]

    book_data.append({
        "Book Name": name,
        "Price": price,
        "Rating": rating
    })

df = pd.DataFrame(book_data)

print("Data extracted successfully!")
print(df.head())
```

STEP 6- Checked scraped data in Dataframe

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

STEP 7- Convert price in numeric value

```python
df["Price"] = df["Price"].str.replace("Â£", "", regex=False).astype(float)

print(df.head())
print("\nPrice Data Type:", df["Price"].dtype)
```

STEP 8- Check dataset duplicate records


```python
print("Duplicate rows:", df.duplicated().sum())

if df.duplicated().sum() > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
else:
    print("No duplicate rows found.")
```

STEP 9- Save Scraped dataset

```python
df.to_csv("books_scraped.csv", index=False)

print("CSV file created successfully!")
print("Total records:", len(df))
```
