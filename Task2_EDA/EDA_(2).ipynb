Step 2-Load the Scraped Dataset

```python
from google.colab import files

uploaded = files.upload()
import pandas as pd

df = pd.read_csv("books_scraped.csv")

print("Dataset loaded successfully!")
display(df.head())
```

STEP 3 — Explore Dataset Structure

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Records:")
display(df.head())
```

STEP 4: Check Missing Values

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

STEP 5: Generate Descriptive Statistics

```python
print("Descriptive Statistics:")

print("\nPrice Statistics:")
print(df["Price"].describe())

print("\nRating Statistics:")
print(df["Rating"].describe())
```

STEP 6: Ask Meaningful Questions

```python
print("Meaningful Questions for EDA:")

questions = [
    "1. What is the average price of the books?",
    "2. What is the minimum and maximum book price?",
    "3. Which rating is most common among the books?",
    "4. Are higher-rated books generally more expensive?",
    "5. Are there any unusual or extreme book prices?"
]

for question in questions:
    print(question)
```

STEP 7: Analyze Trends and Patterns

```python
print("TREND AND PATTERN ANALYSIS")

# Price analysis
print("\nPrice Analysis:")
print("Average Price:", round(df["Price"].mean(), 2))
print("Minimum Price:", df["Price"].min())
print("Maximum Price:", df["Price"].max())

# Rating pattern
print("\nRating Distribution:")
print(df["Rating"].value_counts())

# Most common rating
most_common_rating = df["Rating"].mode()[0]
print("\nMost Common Rating:", most_common_rating)
```

STEP 8: Analyze Relationships

```python
rating_map = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df["Rating_Numeric"] = df["Rating"].map(rating_map)

print("Average Price by Rating:")
print(df.groupby("Rating_Numeric")["Price"].mean())

print("\nCorrelation between Price and Rating:")
print(df["Price"].corr(df["Rating_Numeric"]))
```

STEP 9: Detect Anomalies and Outliers

```python
Q1 = df["Price"].quantile(0.25)
Q3 = df["Price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Price"] < lower_limit) |
    (df["Price"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nNumber of Outliers:", len(outliers))

print("\nOutlier Records:")
display(outliers)
```

STEP 10: Identify Data Issues

```python
print("DATA QUALITY CHECK")

print("\nDuplicate Records:", df.duplicated().sum())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)

print("\nRating Values:")
print(df["Rating"].unique())

print("\nPrice Range:")
print("Minimum Price:", df["Price"].min())
print("Maximum Price:", df["Price"].max())
```

STEP 11: Summarize Key Insights

```python
print("KEY INSIGHTS FROM EDA")
print("-" * 40)

print("1. The dataset contains", df.shape[0], "books.")
print("2. The dataset contains", df.shape[1], "columns.")
print("3. Missing values:", df.isnull().sum().sum())
print("4. Duplicate records:", df.duplicated().sum())
print("5. Average book price:", round(df["Price"].mean(), 2))
print("6. Minimum book price:", df["Price"].min())
print("7. Maximum book price:", df["Price"].max())
print("8. Most common rating:", df["Rating"].mode()[0])
print("9. Number of price outliers:", len(outliers))
```
